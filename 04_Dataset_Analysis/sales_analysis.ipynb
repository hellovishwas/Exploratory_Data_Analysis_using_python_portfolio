# Sales Analysis



### SALES ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={"Product":["A","B","C","A","B","C"],
      "Sales":[100,150,200,120,180,220],
      "Region":["North","South",'East','West',"North","South"]}
df=pd.DataFrame(data)
df
```

```python
Total=df.groupby('Product')["Sales"].sum()
df.groupby('Product')["Sales"].sum().sort_values(ascending=False).head(1)
df[df["Sales"]>150]
plt.subplot(1,2,1)
plt.bar(Total.index,Total.values,color=["green","orange",'red'])
plt.title("Sales by Product")
plt.subplot(1,2,2)
plt.hist(df["Sales"],bins=12)
plt.title("Sales distribution")
plt.tight_layout()
plt.show()
```
