# Students Result Analysis



### STUDENTS RESULT ANALYSIS


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={"Name":["A","B","C","D","E"],
      "Maths":[78,85,96,65,89],
      "Science":[88,79,92,70,95],
      "English":[90,85,88,60,92]}
df=pd.DataFrame(data)
df
```

```python
df["Total"]=df["Maths"]+df["Science"]+df["English"]
df
```

```python
df["Average"]=df["Total"]/3
df
```

```python
df.sort_values("Average",ascending=False).head(1)
```

```python
plt.bar(df['Name'],df['Average'])
plt.xlabel("Students")
plt.ylabel("Average marks")
plt.show()
```

```python
c=np.array([1,2,3,4,5])
plt.figure(figsize=(16,6))
plt.bar(c-0.2,df['Maths'],width=0.2,label="Maths",color="pink",edgecolor='blue')
plt.bar(c+0.2,df['English'],width=0.2,label="English",color="brown",edgecolor='black')
plt.bar(c,df['Science'],width=0.2,label="Science",color="orange",edgecolor='green')
plt.legend(loc="upper right",frameon=True)
plt.title("Subject-wise Marks Comparison",fontweight="bold",fontsize=20)
plt.xlabel("Students",fontsize=16)
plt.ylabel("Marks",fontsize=16)
plt.xticks(c,df["Name"],rotation=30)
plt.show()
```

```python
total=df[["Maths","Science",'English']].sum()
plt.pie(total,labels=["Maths",'English',"Science"],autopct="%1.1f%%")

plt.title("Subjectwise Contribution",fontweight="bold",fontsize=20)
plt.show()
```

```python
plt.pie(df["Average"],labels=df['Name'],autopct="%1.1f%%")
plt.title("Average contribution of Each students")
plt.show()
```
