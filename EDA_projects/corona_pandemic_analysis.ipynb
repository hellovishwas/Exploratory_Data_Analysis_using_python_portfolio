```python
print("TOTAL CASES:",df["Confirmed"].sum())
a=np.array(df.sort_values(by='Death_Rate',ascending=True).head(10)["Country/Region"])
b=np.array(df.sort_values(by='Recovery_Rate',ascending=False).head(10)["Country/Region"])
print("BEST PLACES During CORONA PANDEMIC:",np.intersect1d(a,b))


```

```python
safe_countries=df[(df["Death_Rate"]<2) & (df['Recovery_Rate']>90)]
safe_countries=safe_countries.sort_values(by="Recovery_Rate",ascending=False)
print(safe_countries[["Country/Region","Death_Rate","Recovery_Rate"]].head(10))


```

```python
s=df.sort_values(by="Recovery_Rate",ascending=False).head()
print("Top 5 countries or regions with best recovery rate:\n",s[["Country/Region",'Recovery_Rate']])
```
