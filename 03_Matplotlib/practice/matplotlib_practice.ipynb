Matplotlib

Matplotlib is a Python library used to create graphs and visualizations like line charts, bar charts, pie charts, etc.

```python
import matplotlib.pyplot as plt   #pyplot is the main module used for plotting.
```

### DIFFERENT PLOTS

```python
x=[1,2,3,4,5]        #x = horizontal values
y=[10,20,25,30,60]   #y = vertical values
plt.plot(x,y)        #plot() = creates graph
plt.show()           #show() = displays graph
```

```python
plt.plot(x,y)
plt.title("My First Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.plot(x,y,linestyle="--",marker="o")#linestyle='--' → dashed line
plt.title("Styled Graph")#marker='o' → points shown
plt.show
```

```python
x=[1,2,3,4]
y1=[10,20,30,40]
y2=y1[::-1]
plt.plot(x,y1,label="line 1")
plt.plot(x,y2,label="line 2")
plt.legend()#It creates a small box on the graph that tells:which line represents what
plt.show()
```

```python
names=["A","B","C"]
values=[5,7,3]
plt.bar(names,values)
plt.title("Bar Chart")
plt.show()
```

```python
labels=["Python","Java","C++"]
sizes=[50,30,20]
plt.pie(sizes,labels=labels,autopct="%1.1f%%")#autopct=this is used inside plt.pie() to display percentages on the pie chart.
plt.title("Pie Chart")
plt.show()
```

```python
x=[10,20,30,40]
y=[1,3,5,7]
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.show()
```

### RESULT

```python
x=["Maths","English","Hindi","Science","Social"]
y=[18,29,16,14,20]
plt.plot(x,y,linestyle="--",marker="o")
plt.title("Result")
plt.xlabel("Marks")
plt.ylabel("Subjects")
plt.show()
```

```python
subjects=["Maths","English","Hindi","Science","Social"]
marks=[18,29,16,14,20]
```

```python
avg=sum(marks)/len(subjects)
max_marks=max(marks)
top_subject=subjects[marks.index(max_marks)]
```

```python
plt.figure(figsize=(10,5))
```

```python
plt.subplot(1,1,1)
plt.bar(subjects,marks)
plt.title("Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")
```

```python
plt.subplot(1,2,2)
plt.plot(subjects,marks,marker="o")
plt.title("Marks Trend")
```

```python
plt.axhline(avg,linestyle=":",label=f"Average={avg:.2f}")#axhline:Draw a horizontal line across the graph
plt.legend()
```

```python
index = marks.index(max_marks)
plt.plot(subjects, marks, marker='o')
plt.text(subjects[index], max_marks + 1,
         f"Top: {top_subject}",
         ha="center")
plt.title("Result")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

### SUBPLOT AND SUBPLOTS AND GRIDS

```python
x=[1,2,3,4]
y=[10,20,30,40]
fig,ax=plt.subplots()
ax.plot(x,y)
ax.set_title("My Graph")
ax.set_xlabel("X axis")
ax.set_ylabel("Y axis")
plt.show()
```

```python
ax.plot(x,y,label="Line 1")
ax.plot(x,[5,10,15,20],label="Line 2")
ax.legend()
```

```python
ax.bar(x,y)
ax.scatter(x,y)
```

```python
ax.axhline(20)
```

```python
ax.text(2,20,"Point")
```

```python
ax.grid()
```

```python
fig,ax=plt.subplots(2,1)
ax[0].plot(x,y)
ax[0].set_title("Line")
ax[1].bar(x,y)
ax[1].set_title("Bar")
plt.show()
```

```python
plt.plot(x,y,color="red")
```

```python
plt.plot(x,y,linewidth=3,linestyle=":",marker="p")
```
