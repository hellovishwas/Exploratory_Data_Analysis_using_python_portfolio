Merging joining and concatanation

```python
employees=pd.DataFrame({
    "employee_id":[1,2,3,7,8],
    "name":["John","Anna","Peter","Linda","Bob"],
    "department":["HR","IT","Finance","IT",'HR']
    })
```

```python
salaries=pd.DataFrame({"employee_id":[1,2,3,4,5],
                       'salary':[6000,8000,6500,7000,9000],
                       'bonus':[5000,1000,7000,8000,12000]
                       })
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on="employee_id",how="inner")#this will merge the dataframes on the basis of common employee id
```

```python
pd.merge(employees,salaries,on="employee_id",how="outer")#this will merge the dataframes
```

```python
pd.merge(employees,salaries,on="employee_id",how="left")#this will merge the dataframes if names are given
```

```python
pd.merge(employees,salaries,on="employee_id",how="right")#this will merge the dataframes on the basis of salary given
```

Group by aggregation

```python
data={
    "Category":['A','B','A','B','A','B'],
    "Store":['S1','S2','S2','S2','S1','S1'],
    "Sales":[100,150,200,250,300,350],
    "Quantity":[10,15,12,18,8,20],
    "Date":[10,15,12,18,8,20],
}
df=pd.DataFrame(data)
df
```

```python
#group by category and calculate the sum of sales
df.groupby("Category")['Sales'].sum()
```

```python
#group by store and calculate the sum of sales

df.groupby("Store")['Sales'].sum()
```

```python
#group by multiple columns
#group by category and store
df.groupby(["Category",'Store'])["Sales"].sum()
```

Aggregation

```python
df["Sales"].count()
#median,mean,mode,min ,max ,count,std deviation
```

```python
df["Sales"].mode()#it gives full table bcz all values having same number
```

```python
df["Sales"].agg(["mean","median","max"])#mode is not working here bcz it gives full tble if the value of mode is more than one
```

Pivot tables

```python
data = {
    "Date": pd.date_range("2023-01-01", periods=20),
    "Product": ["A", "B", "C", "D"] * 5,
    "Region": ["East", "West", "North", "South",
               "East", "West", "North", "South",
               "East", "West", "North", "South",
               "East", "West", "North", "South",
               "East", "West", "North", "South"],
    "Sales": np.random.randint(100, 1000, 20),
    "Units": np.random.randint(10, 100, 20),
    "Rep": ["John", "Mary", "Bob", "Alice",
            "John", "Mary", "Bob", "Alice",
            "John", "Mary", "Bob", "Alice",
            "John", "Mary","Bob", "Alice",
            "John", "Mary","Bob", "Alice",]
}
df=pd.DataFrame(data)
df["Month"]=df["Date"].dt.month_name()
df["Quarter"]="Q"+df["Date"].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='Sales',index="Region",columns="Product",aggfunc="median")#default aggregation function is mean
```

```python
pivot=pd.pivot_table(df,values=["Sales","Units"],index="Region",columns="Product")
pivot#values must be numeric
```

cross tab

```python
pd.crosstab(df["Region"],df["Product"])#defaultly gives count
```
