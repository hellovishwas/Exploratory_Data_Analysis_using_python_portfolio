### STUDENTS INFO

```python
data={'Name':["A","B",'C','D',"E",'F','G','H'],
      'Maths':[80,90,np.nan,70,60,85,90,np.nan],
      'Science':[75,85,80,np.nan,65,88,92,70],
      "English":[70,80,75,60,np.nan,82,85,78],
      "City":["Delhi",'Mumbai',"Delhi",'Mumbai',"Delhi","Pune",'Delhi',"Indore"]
      }
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.info())
print(df.describe())
```

```python
df['Maths']=df['Maths'].fillna(df["Maths"].mean())
df['English']=df['English'].fillna(df["English"].mean())
df["Science"]=df["Science"].fillna(df['Science'].mean())
df['Total']=df['Maths']+df["Science"]+df["English"]
```

```python
df['Average'] = df[["Maths","Science","English"]].mean(axis=1)
df
```

```python
df['Grade']=df['Average'].apply(lambda x: 'A' if x>=85 else ('B' if x>=70 else 'C'))
```

```python
print(df.groupby('City')['Average'].mean())
```

```python
print(df.sort_values("Average",ascending=False).head(1))
```

```python
print(df[df['Average']>80])
```

```python
arr=df[["Maths","Science","English"]]
print(arr)
```

```python
print(np.mean(arr))
```

```python
print(np.max(arr,axis=0))
```

```python
df.sort_values("Average",ascending=False).tail(1)
```

```python
df.groupby('City')["Average"].mean().idxmax()
```

```python
df["pass/fail"]=df['Average'].apply(lambda x: "pass" if x>=70  else "fail")
df
```

```python
df[["Maths","Science","English"]].corr()
```

```python
corr=df.select_dtypes(include="number").corr()["Average"]
```

```python
corr = df.select_dtypes(include='number').corr()

f=corr["Average"].sort_values(ascending=False)
```

```python
pd.DataFrame(f)
f
print(f[f<1].head(1))
```
