```python
##valid sudoku solution
sudoku=np.array([[5,3,4,6,7,8,9,1,2],
                 [6,7,2,1,9,5,3,4,8],
                 [1,9,8,3,4,2,5,6,7],
                 [8,5,9,7,6,1,4,2,3],
                 [4,2,6,8,5,3,7,9,1],
                 [7,1,3,9,2,4,8,5,6],
                 [9,6,1,5,3,7,2,8,4],
                 [2,8,7,4,1,9,6,3,5],
                 [3,4,5,2,8,6,1,7,9]])
```

```python

a=np.sum(sudoku,axis=1)
a
```

```python

b=np.sum(sudoku,axis=0)
b
```

```python
if np.all(a!=45) or np.all(b!=45):
    print("This is invalid sudoku")

else:print("This is valid sudoku")
```

```python
def check_3x3_boxes(sudoku):
    for i in range(0,9,3):
        for j in range(0,9,3):
            box_sum=np.sum(sudoku[i:i+3,j:j+3])
            if box_sum!=45:
                print("invalid sudoku")
                break
    else:
        print("valid sudoku")
check_3x3_boxes(sudoku)
```
